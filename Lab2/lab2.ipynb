Для кожного розв'язку виведіть: оптимальний план перевезень, сумарну вартість

```python
c_flat = D.flatten()

A_eq = []
b_eq = []
m = 3
n =4

for i in range(m):
    row = np.zeros((m, n))
    row[i, :] = 1
    A_eq.append(row.flatten())
    b_eq.append(запас[i])


for j in range(n):
    row = np.zeros((m, n))
    row[:, j] = 1
    A_eq.append(row.flatten())
    b_eq.append(попит[j])

bounds = [(0, None) for _ in range(m * n)]

res = linprog(c_flat, A_eq=np.array(A_eq), b_eq=np.array(b_eq), bounds=bounds, method='highs')

plan_scipy = res.x.reshape((m, n))

print("\n" + "="*50)
print(" РОЗВ'ЯЗОК 2: SciPy (HiGHS)")
print("="*50)
print("Оптимальний план перевезень:")
print(pd.DataFrame(plan_scipy, index=постачальники, columns=споживачі))
print(f"\nЗагальна вартість: {res.fun:.2f}")


dual_eq = res.eqlin.marginals

print("\nТіньові ціни (SciPy HiGHS):")
print("Запас (постачальники):", {постачальники[i]: dual_eq[i] for i in range(m)})
print("Попит (споживачі):    ", {споживачі[j]: dual_eq[m + j] for j in range(n)})
```

```python
import numpy as np
import pandas as pd
from ortools.linear_solver import pywraplp

постачальники = ["Басівка", "Київ", "Донецьк"]
споживачі = ["Дніпро", "Миколаїв", "Татарів", "Рівне"]

D = np.array([
    [1005, 875, 220, 215],
    [450, 480, 680, 330],
    [250, 490, 1280, 930]
])

запас = [120, 60, 30]
попит = [70,100,20,20] 

m, n = D.shape

solver = pywraplp.Solver.CreateSolver('GLOP')

x = {}
for i in range(m):
    for j in range(n):
        x[i, j] = solver.NumVar(0, solver.infinity(), f'x_{i}_{j}')

supply_constraints = []
for i in range(m):
    constraint = solver.Constraint(запас[i], запас[i], f'Supply_{постачальники[i]}')
    for j in range(n):
        constraint.SetCoefficient(x[i, j], 1)
    supply_constraints.append(constraint)

demand_constraints = []
for j in range(n):
    constraint = solver.Constraint(попит[j], попит[j], f'Demand_{споживачі[j]}')
    for i in range(m):
        constraint.SetCoefficient(x[i, j], 1)
    demand_constraints.append(constraint)

objective = solver.Objective()
for i in range(m):
    for j in range(n):
        objective.SetCoefficient(x[i, j], float(D[i, j]))
objective.SetMinimization()

status = solver.Solve()

if status == pywraplp.Solver.OPTIMAL:
    plan_ortools = np.zeros((m, n))
    for i in range(m):
        for j in range(n):
            plan_ortools[i, j] = x[i, j].solution_value()

    print("=== РОЗВ'ЯЗОК 1: Google OR-Tools ===")
    print("Оптимальний план перевезень:")
    print(pd.DataFrame(plan_ortools, index=постачальники, columns=споживачі))
    print(f"\nЗагальна вартість: {objective.Value():.2f}")

    print("\nТіньові ціни:")
    print("Запас (постачальники):", {постачальники[i]: supply_constraints[i].dual_value() for i in range(m)})
    print("Попит (споживачі):    ", {споживачі[j]: demand_constraints[j].dual_value() for j in range(n)})
```
